```python
## Q1. Explain Class and Object with respect to Object-Oriented Programming. Give a suitable example.
# a class is a blueprint, while an object is a real instance created from that blueprint.
class student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
         
    def return_student_details(self): # Class → Object → __init__ → Object data
        return self.name, self.age
```

```python
aman = student("aman",23) # aman-object, student-class
```

```python
aman.name 
```

```python
aman.return_student_details() 
```

```python
# Q2. Name the four pillars of OOPs.
# Encapsulation — Wrapping data and methods together inside a class.
# Inheritance — Creating a new class from an existing class and reusing its properties/methods.
# Polymorphism — One method or function can behave differently depending on the object.
# Abstraction — Hiding unnecessary implementation details and showing only the important features.
```

```python
# Q3. Explain why the __init__() function is used. Give a suitable example.
# The __init__() function is a special method in Python that is used to initialize the attributes of an object when the object is created.
class school:
    def __init__(self,name,age):
        self.name = name
        self.age = age
student1= school("aman",20)
print(student1.name)
print(student1.age) 
```

```python
# Q4. Why self is used in OOPs?
# self is used to refer to the current object inside a class.

# It allows us to access the object's attributes and methods.
class school:
    def __init__(self,name,age):
        self.name = name
        self.age = age
student1= school("aman",20)
print(student1.name)
print(student1.age) 
```

```python
# Q5. What is inheritance? Give an example for each type of inheritance.

#Inheritance is the process by which one class gets the properties and methods of another class.

# Types of Inheritance
# Single Inheritance
# One parent → One child
# Example: Animal → Dog

# Multiple Inheritance
# Multiple parents → One child
# Example: Father + Mother → Child

# Multilevel Inheritance
# Parent → Child → Grandchild
# Example: Grandfather → Father → Son

# Hierarchical Inheritance
# One parent → Multiple children
# Example: Animal → Dog and Animal → Cat

# Hybrid Inheritance
# Combination of two or more types of inheritance.
```
