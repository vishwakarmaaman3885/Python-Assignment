```python
# Q3. Remove duplicates from the given list using a set.
List = [1, 1, 1, 2, 1, 3, 1, 4, 2, 1, 2, 2, 2, 3, 2, 4, 3, 1, 3, 2, 3, 3, 3, 4, 4, 1, 4, 2, 4, 3, 4, 4]
unique_list = set(List) # or if you  got answer on list(set(List))
print(unique_list) 
```

```python
# Q4. Demonstrate union() and update() methods with code examples.
set1 = {1, 2, 3}
set2 = {3, 4, 5}
result = set1.union(set2) # union-Combines two sets and returns a new set.
print(result)
print(set1)
```

```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}
result = set1.update(set2)
print(result)
print(set1)  # Modifies the original set
```

```python
# Q6. Create a nested dictionary.
student = {"Name":{"Aman","Raj","Roshini","Madhu"}, "Course_name":("It","Cyber","Data science")}  
print(student) 
```

```python
# Q7. Use setdefault() to add the key topics.
dict1 = {'language':'Python','course':'Data science masters'}
dict1.setdefault('topics',['python','machine learning','deep learning'])  # adding something default value
print(dict1) 
```

```python
# Q8. What are the three view objects in dictionaries? Use the three in-built methods in python to display
#  these three view objects for the given dictionary.
# dict1 = {'Sport': 'Cricket' , 'Teams': ['India', 'Australia', 'England', 'South Africa', 'Sri Lanka', 'New Zealand']}
dict1 = {"Sport":'Cricket','Teams':['India',"Australia",'England','South Africa','Sri Lanka','New Zealand']} 
print(dict1.keys())
print(dict1.values())
print(dict1.items()) 
```
